```python
import json
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
df=pd.read_csv("online retail clean.csv")
df.head()
```

```python
df=df.drop(columns="Unnamed: 0")
```

```python
df = df.dropna(subset=["Customer ID"])
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]
```

```python
df.head()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
cut_date=df["InvoiceDate"].max() - pd.DateOffset(months=3)
train_df=df[df["InvoiceDate"]<=cut_date]
future_df=df[df["InvoiceDate"]>cut_date]
print(f"Cutoff date: {cut_date}")
print(f"Feature window: {train_df['InvoiceDate'].min()} to {train_df['InvoiceDate'].max()}")
print(f"Target window: {future_df['InvoiceDate'].min()} to {future_df['InvoiceDate'].max()}")
```

```python
snapshot_date = cut_date + pd.Timedelta(days=1)

customer_features = train_df.groupby("Customer ID").agg(
    recency=("InvoiceDate", lambda x: (snapshot_date - x.max()).days),
    frequency=("Invoice", "nunique"),
    monetary_total=("Revenue", "sum"),
    monetary_avg=("Revenue", "mean"),
    avg_basket_size=("Quantity", "mean"),
    tenure=("InvoiceDate", lambda x: (snapshot_date - x.min()).days),
    product_diversity=("StockCode", "nunique"),
).reset_index()

customer_features["spend_per_tenure_day"] = (
    customer_features["monetary_total"] / customer_features["tenure"].replace(0, 1)
)
customer_features["purchase_rate_per_month"] = (
    customer_features["frequency"] / (customer_features["tenure"].replace(0, 1) / 30)
)

customer_features.head()
```

```python
target = future_df.groupby("Customer ID")["Revenue"].sum().reset_index()
target = target.rename(columns={"Revenue": "future_spend"})

print(f"Customers with future purchases: {target.shape[0]}")
target.head()
```

```python

model_df = customer_features.merge(target, on="Customer ID", how="left")
model_df["future_spend"] = model_df["future_spend"].fillna(0)

print(f"Final modeling dataset shape: {model_df.shape}")
print(f"Customers with $0 future spend (churned/lapsed): {(model_df['future_spend'] == 0).sum()} "
      f"({(model_df['future_spend'] == 0).mean():.1%})")

model_df.head()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(model_df["future_spend"], bins=50)
axes[0].set_title("future_spend (raw)")
axes[0].set_xlabel("Future spend ($)")

axes[1].hist(np.log1p(model_df["future_spend"]), bins=50)
axes[1].set_title("future_spend (log1p transformed)")
axes[1].set_xlabel("log(1 + spend)")

plt.tight_layout()
plt.show()

model_df["future_spend"].describe()
```

```python
cohort_df = df.copy()
cohort_df["OrderMonth"] = cohort_df["InvoiceDate"].dt.to_period("M")
cohort_df
```

```python
first_purchase = cohort_df.groupby("Customer ID")["OrderMonth"].min().reset_index()
first_purchase.columns = ["Customer ID", "CohortMonth"]
first_purchase
```

```python
cohort_df = cohort_df.merge(first_purchase, on="Customer ID")
cohort_df["CohortIndex"] = (
    (cohort_df["OrderMonth"].dt.year - cohort_df["CohortMonth"].dt.year) * 12
    + (cohort_df["OrderMonth"].dt.month - cohort_df["CohortMonth"].dt.month)
)

```

```python
cohort_df
```

```python
cohort_counts = cohort_df.groupby(["CohortMonth", "CohortIndex"])["Customer ID"].nunique().reset_index()
cohort_pivot = cohort_counts.pivot(index="CohortMonth", columns="CohortIndex", values="Customer ID")
cohort_pivot
```

```python
cohort_size = cohort_pivot.iloc[:, 0]
retention = cohort_pivot.divide(cohort_size, axis=0)

retention.round(3)
import seaborn as sns

plt.figure(figsize=(14, 8))
sns.heatmap(retention, annot=True, fmt=".0%", cmap="Blues")
plt.title("Customer Retention by Cohort")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Cohort (Acquisition Month)")
plt.show()
```

```python
# ==================================================
# Block 4: Two-Stage Model — Setup
# ==================================================

from sklearn.model_selection import train_test_split

# Define feature columns (everything except IDs and target)
feature_cols = [
    "recency", "frequency", "monetary_total", "monetary_avg",
    "tenure", "avg_basket_size", "product_diversity",
    "spend_per_tenure_day", "purchase_rate_per_month"
]

# Stage 1 target: did this customer purchase at all in the future window?
model_df["will_purchase"] = (model_df["future_spend"] > 0).astype(int)

X = model_df[feature_cols]
y_class = model_df["will_purchase"]      # Stage 1 target
y_reg = np.log1p(model_df["future_spend"])  # Stage 2 target (log-transformed)

# Single split, shared across both stages — same customers in train/test throughout
X_train, X_test, y_class_train, y_class_test, y_reg_train, y_reg_test = train_test_split(
    X, y_class, y_reg, test_size=0.2, random_state=42, stratify=y_class
)

print(f"Train size: {X_train.shape[0]}, Test size: {X_test.shape[0]}")
print(f"Train purchase rate: {y_class_train.mean():.1%}")
print(f"Test purchase rate: {y_class_test.mean():.1%}")
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report, roc_auc_score

# Define the hyperparameter grid to search over
param_grid = {
    "n_estimators": [50,100, 200],
    "learning_rate": [0.01,0.05, 0.1],
    "max_depth": [2, 3, 4],
    "min_samples_leaf": [5,10, 20],
}

base_clf = GradientBoostingClassifier(random_state=42)

grid_search = GridSearchCV(
    estimator=base_clf,
    param_grid=param_grid,
    scoring="roc_auc",  
    cv=5,                
    n_jobs=-1,             
    verbose=1
)

grid_search.fit(X_train, y_class_train)

print(f"Best params: {grid_search.best_params_}")
print(f"Best CV ROC-AUC: {grid_search.best_score_:.3f}")

clf = grid_search.best_estimator_
y_class_pred = clf.predict(X_test)
y_class_proba = clf.predict_proba(X_test)[:, 1]

print(classification_report(y_class_test, y_class_pred))
print(f"Test ROC-AUC: {roc_auc_score(y_class_test, y_class_proba):.3f}")
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


train_mask = y_class_train == 1
test_mask = y_class_test == 1

X_train_reg = X_train[train_mask]
y_train_reg = y_reg_train[train_mask]   

X_test_reg = X_test[test_mask]
y_test_reg = y_reg_test[test_mask]

print(f"Stage 2 train size: {X_train_reg.shape[0]} (purchasers only)")
print(f"Stage 2 test size: {X_test_reg.shape[0]} (purchasers only)")
```

```python
param_grid_reg = {
    "n_estimators": [100, 200],
    "learning_rate": [0.05, 0.1],
    "max_depth": [2, 3, 4],
    "min_samples_leaf": [10, 20],
}

base_reg = GradientBoostingRegressor(random_state=42)

grid_search_reg = GridSearchCV(
    estimator=base_reg,
    param_grid=param_grid_reg,
    scoring="neg_mean_absolute_error",  
    cv=5,
    n_jobs=-1,
    verbose=1
)

grid_search_reg.fit(X_train_reg, y_train_reg)

print(f"Best params: {grid_search_reg.best_params_}")
print(f"Best CV MAE (log scale): {-grid_search_reg.best_score_:.3f}")

reg = grid_search_reg.best_estimator_
```

```python
y_pred_log = reg.predict(X_test_reg)

mae_log = mean_absolute_error(y_test_reg, y_pred_log)
r2_log = r2_score(y_test_reg, y_pred_log)

y_test_dollars = np.expm1(y_test_reg)
y_pred_dollars = np.expm1(y_pred_log)

mae_dollars = mean_absolute_error(y_test_dollars, y_pred_dollars)

print(f"Log-scale  — MAE: {mae_log:.3f}, R²: {r2_log:.3f}")
print(f"Real-dollar — MAE: ${mae_dollars:.2f}")
```

```python
purchase_proba = clf.predict_proba(X_test)[:, 1]           # Stage 1: P(will purchase)
predicted_spend_log = reg.predict(X_test)                   # Stage 2: predicted spend IF they purchase
predicted_spend_dollars = np.expm1(predicted_spend_log)

# Expected value = probability × conditional spend
expected_clv = purchase_proba * predicted_spend_dollars

results = pd.DataFrame({
    "Customer ID": model_df.loc[X_test.index, "Customer ID"].values,
    "actual_future_spend": np.expm1(y_reg_test).values,
    "purchase_probability": purchase_proba,
    "predicted_spend_if_buying": predicted_spend_dollars,
    "expected_clv": expected_clv
})

results.sort_values("expected_clv", ascending=False).head(10)
```

```python
import shap

# TreeExplainer is built specifically for tree-based models (fast, exact)
explainer_clf = shap.TreeExplainer(clf)
shap_values_clf = explainer_clf.shap_values(X_test)

# Summary plot: shows which features matter most, and in which direction
shap.summary_plot(shap_values_clf, X_test, feature_names=feature_cols)
```

```python
explainer_reg = shap.TreeExplainer(reg)
shap_values_reg = explainer_reg.shap_values(X_test_reg)

shap.summary_plot(shap_values_reg, X_test_reg, feature_names=feature_cols)
```

```python
# Split customers into tiers using quartiles of expected_clv
results["clv_tier"] = pd.qcut(
    results["expected_clv"],
    q=[0, 0.25, 0.5, 0.75, 1.0],
    labels=["Low", "Medium", "High", "Top"]
)

# Attach RFM profile back onto the results table for cross-tabulation
segment_profile = results.merge(
    model_df.loc[X_test.index, ["Customer ID"] + feature_cols],
    on="Customer ID"
)

# Average RFM profile per tier — this is the "what do our best customers look like" table
tier_summary = segment_profile.groupby("clv_tier")[feature_cols + ["expected_clv"]].mean().round(2)
tier_summary
```

```python
at_risk = segment_profile[
    (segment_profile["monetary_total"] > segment_profile["monetary_total"].median()) &
    (segment_profile["purchase_probability"] < 0.3)
].sort_values("monetary_total", ascending=False)

print(f"At-risk high-value customers: {at_risk.shape[0]}")
at_risk[["Customer ID", "monetary_total", "purchase_probability", "expected_clv"]].head(10)
```

```python

from sklearn.metrics import roc_curve, auc

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Predicted vs Actual (regressor, purchasers only, dollar scale)
axes[0, 0].scatter(y_test_dollars, y_pred_dollars, alpha=0.4)
max_val = max(y_test_dollars.max(), y_pred_dollars.max())
axes[0, 0].plot([0, max_val], [0, max_val], "r--")  # perfect-prediction line
axes[0, 0].set_xlabel("Actual future spend ($)")
axes[0, 0].set_ylabel("Predicted spend ($)")
axes[0, 0].set_title("Predicted vs Actual Spend (purchasers)")

# 2. ROC Curve (classifier)
fpr, tpr, _ = roc_curve(y_class_test, y_class_proba)
roc_auc = auc(fpr, tpr)
axes[0, 1].plot(fpr, tpr, label=f"AUC = {roc_auc:.3f}")
axes[0, 1].plot([0, 1], [0, 1], "r--")  # random-guess baseline
axes[0, 1].set_xlabel("False Positive Rate")
axes[0, 1].set_ylabel("True Positive Rate")
axes[0, 1].set_title("ROC Curve — Purchase Classifier")
axes[0, 1].legend()

# 3. CLV distribution across the customer base
axes[1, 0].hist(results["expected_clv"], bins=50)
axes[1, 0].set_xlabel("Expected CLV ($)")
axes[1, 0].set_title("Expected CLV Distribution")

# 4. Top-10 customers table rendered as a bar chart
top10 = results.sort_values("expected_clv", ascending=False).head(10)
axes[1, 1].barh(top10["Customer ID"].astype(str), top10["expected_clv"])
axes[1, 1].set_xlabel("Expected CLV ($)")
axes[1, 1].set_title("Top 10 Customers by Expected CLV")
axes[1, 1].invert_yaxis()

plt.tight_layout()
plt.show()
```
